```python
# 2D SIMPLE TRAJECTORY ANALYSIS # 

import math 


 
def graph_plotter():

    y_max = 40.00  # fill here
    y_slots = 2000  # fill here

    j_max = int((y_max/10)*(y_slots))
    j_step = int(y_max/10)
    j_scaledown = 1/(y_slots/10)

    
    x_max = 20.00  # fill here
    x_slots = 1000  # fill here

    i_max = int((x_max/10)*(x_slots)) + 1
    i_step = int(x_max/10)
    i_scaledown = 1/(x_slots/10)

    
    resolution = y_slots*x_slots
    dp = int(math.log((x_slots/10),10)) # could put 1/i_scaledown in place of x_slots/10
    tolerance = x_max/x_slots

##################################################
    with open("Graph_File.txt","w") as import_file:

   
      for j in range(j_max,-1,-j_step): 
      
         y = j*j_scaledown 
      
         for i in range(0,i_max,i_step): 

            reference = 0
            taylor = 0 
      
            x = i*i_scaledown

         
            fofx = (15-x)*(math.sin(x-0.8)) + 20

            pofx = 30.2662 + ((7.7161)*(x-1.68))/(1) + ((-11.5405)*(x-1.68)**2)/(2) + ((-6.1746)*(x-1.68)**3)/(math.factorial(3)) + ((12.8148)*(x-1.68)**4)/(math.factorial(4)) + ((4.6331)*(x-1.68)**5)/(math.factorial(5)) + ((-14.0891)*(x-1.68)**6)/(math.factorial(6)) + (-3.0917)*(x-1.68)**7/(math.factorial(7))


            if y == fofx: 
               reference += 1 

            if ( y != fofx and abs(y - round(fofx,dp)) <= tolerance ): 
               reference += 1 

            if ( y != round(fofx,dp) and abs(y - round(fofx,dp)) > tolerance ): 
               reference += 0 

      
            if y == pofx: 
               taylor += 1 
            
            if y != pofx and abs(y - round(pofx,dp)) <= tolerance : 
               taylor += 1 

            if ( y != round(pofx,dp) and abs(y - round(pofx,dp)) > tolerance ): 
               taylor += 0 
            

            if reference == taylor and reference > 0: 
               import_file.write("OO")

            if reference == taylor and reference == 0: 
               import_file.write("  ")

            if reference > taylor: 
               import_file.write("RR")

            if reference < taylor:
               import_file.write("TT")

             
            
            if str(x) == str(x_max):
               import_file.write("\n")

    return "Graph Resolution:",resolution 

            
print(graph_plotter())


```
