This version is both the generalized version and specialized in generating graphs for 1:1 continous functions. It should be noted that you would need to install numpy onto your source code editor to run this. 

```python
import numpy 

def graph_gen(): 
    

    y_min = -20.00 
    y_max = 20.00  
    y_points = 4001  
   


    x_min = -20.00 
    x_max = 20.00  
    x_points = 4001  
    

    
    resolution = y_points*x_points

   

 ##################################################
    with open("Graph_File.dat","w") as import_file:


                     
      for y in numpy.linspace(y_max,y_min,y_points): 
      
      
         for x in numpy.linspace(x_min,x_max,x_points): 

   

   

         
            if y == x**2:
               import_file.write("AA")


            if y != x**2: 
               import_file.write("  ")
               
            
            if str(x) == str(x_max):
               import_file.write("\n")

    return "Graph Resolution:",resolution 


print(graph_gen())
```
