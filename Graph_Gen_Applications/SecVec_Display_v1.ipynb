```python
# SECVEC DISPLAY # 

import math 
import copy 
import numpy
 

# all hail .atan2(,) 


vector = [1.00,1.00]
sample_string = "Arvyn Jonathan Singh"




def rotation(vector,rotation_theta,secvec_coords):  

   theta_step = 0.01
   
   c = 0 
   while c < rotation_theta/theta_step:

      radians_theta_step = math.radians(theta_step)

      x1 = vector[0]
      y1 = vector[1]

      x2 = (x1*math.cos(radians_theta_step)) - (y1*math.sin(radians_theta_step))
      y2 = (x1*math.sin(radians_theta_step)) + (y1*math.cos(radians_theta_step))

      vector[0] = x2 
      vector[1] = y2

      pair = [] 
      pair.append(round(vector[0],2))
      pair.append(round(vector[1],2))
      secvec_coords.append(tuple(pair))

      c += 1 

   return secvec_coords 


def extension(vector,secvec_coords): 

   x1 = vector[0]
   y1 = vector[1] 

   # extension of M1 could be 1/20th the absolute length of the x-axis.. as a rule of thumb. May change depending on future outputs 
   M1 = (x1**2 + y1**2)**0.5 
   M2 = M1 + 0.5 # globalize the axis ranges and insertion #, so we can add our graph step to M1 to make M2 
   angle = math.atan2(y1,x1) # in radians already; angle stays same after extension, x,y change accordingly, in response. 

   x2 = M2*math.cos(angle)
   y2 = M2*math.sin(angle)

   vector[0] = x2 
   vector[1] = y2 

   pair = [] 
   pair.append(round(vector[0],2))
   pair.append(round(vector[1],2))
   secvec_coords.append(tuple(pair))

   return secvec_coords
   

def secvec(sample_string,vector): 

   secvec_coords = [] 

   conversion_dict = { "e": 14.00, "t": 65.00, "a": 6.00, "o": 70.00, "i": 13.00, "n": 55.00, "s": 8.00, "h": 80.00, "r": 5.00, "d": 60.00, 
                      "l": 7.00, "c": 180.00, "u": 12.00, "m": 50.00, "w": 9.00, "f": 85.00, "g": 11.00, "y": 45.00, "p": 10.00, "b": 62.00, 
                      "v": 17.00, "k": 78.00, "T": 16.00, "I": 52.00, "A": 19.00, "j": 88.00, "x": 24.00, "q": 40.00, "z": 36.00, "S": 68.00, 
                      "O": 41.00, "C": 58.00, "M": 30.00, "F": 72.00, "P": 112.00, "W": 140.00, "B": 200.00, "E": 48.00, "H": 210.00, "D": 164.00, 
                      "L": 5.00, "R": 176.00, "N": 102.00, "G": 184.00, "U": 170.00, "Y": 230.00, "V": 240.00, "K": 120.00, "J": 130.00, "Q": 150.00, 
                      "X": 280.00, "Z": 0.00, " ": 360.00}


   for character in sample_string: 

      rotation_theta = conversion_dict[character] # here for readability; could chuck this straight into as parameter

      rotation(vector,rotation_theta,secvec_coords)
      extension(vector,secvec_coords)

   return secvec_coords 






def graph_gen(): 

    sec_coords_list = copy.deepcopy(secvec(sample_string,vector))

    print(sec_coords_list)

    sec_coords = set(sec_coords_list)

    print(sec_coords)

    # need to evaluate if to +1 or -1 based on signage of y_min and x_max 


    y_min = -20.00 # fill here
    y_max = 20.00  # fill here
    y_points = 4001  # fill here (axis length/desired step) + 1 
    y_step = (abs(y_max) + abs(y_min))/(y_points)


    x_min = -20.00 # fill here 
    x_max = 20.00  # fill here
    x_points = 4001  # fill here (axis length/desired step) + 1 
    x_step = (abs(x_max) + abs(x_min))/(x_points)

    
    resolution = y_points*x_points

    # Not needed right now # 
    dp = int(math.log((x_points/10),10)) # could put 1/i_scaledown in place of x_slots/10 
    tolerance = x_max/x_points

 ##################################################
    with open("Graph_File.dat","w") as import_file:

                     
      for y in numpy.linspace(y_max,y_min,y_points): 
      
      
         for x in numpy.linspace(x_min,x_max,x_points): 

   

            pair = []
            pair.append(round(x,2))
            pair.append(round(y,2))
            xy_pair = tuple(pair) 

      

         
            if xy_pair in sec_coords:
               import_file.write("SV")


            if xy_pair not in sec_coords: 
               import_file.write("**")
               
            
            if str(x) == str(x_max):
               import_file.write("\n")

    return "Graph Resolution:",resolution 


print(graph_gen())

        






```
