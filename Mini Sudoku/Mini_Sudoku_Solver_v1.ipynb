```python
# 6x6 MINI SUDOKU SOLVER v1 # 

import random 
import copy


def grid_populator(grid):

    digits = [1,2,3,4,5,6]

    for j in range(0,6): # selected a row 

        chosen_already = [] 
        for number in grid[j]:
            if number != 0:
                chosen_already.append(number)

        for i in range(0,6): # selected a slot in row 
            if grid[j][i] == 0: # if a blank slot 
                replacement = random.choice(digits)
                while replacement in chosen_already:
                    replacement = random.choice(digits)
                grid[j][i] = replacement
                chosen_already.append(replacement)

    return grid


def column_checker(grid):

    cc_penalty = 0 

    digits = [1,2,3,4,5,6]


    all_numbers = []
    for j in range(0,6): # selects a row 
        for i in range(0,6): # selects a slot
            all_numbers.append(grid[j][i]) 


    for s in range(0,6): # starter digit 

        sampled_column = []
        for d in range(s,36,6): # index in all_numbers
            sampled_column.append(all_numbers[d])

        for digit in digits:
            if digit not in sampled_column:
                cc_penalty += 1 

    return cc_penalty 


def section_checker(grid):

    sc_penalty = 0

    digits = [1,2,3,4,5,6]


    for j in range(0,5,2): # selects a section

        section = grid[j][0:3] + grid[j+1][0:3]

        for digit in digits:
            if digit not in section:
                sc_penalty += 1 
 
    for j in range(0,5,2):

        section = grid[j][3:6] + grid[j+1][3:6]

        for digit in digits:
            if digit not in section:
                sc_penalty += 1 


    return sc_penalty 


# STOCHASTIC GRID GENERATION MECHANICS ^^^ #
############################################


def cycling_mechanism(grid):

    digits = [1,2,3,4,5,6]

    columns = []
    all_numbers = []

    for row in grid:
        for number in row:
            all_numbers.append(number)
 

    for s in range(0,6):
        column = []
        for d in range(s,36,6):
            column.append(all_numbers[d])
        columns.append(column)
            

    for digit in digits: 

        #Left Sections:
        for j in range(0,6,2): 
            potential_slots = []
            section = grid[j][0:3] + grid[j+1][0:3]

            for i in range(0,3):

                if grid[j][i] == 0:

                    if digit not in grid[j] and digit not in columns[i] and digit not in section:
                        potential_slots.append(j)
                        potential_slots.append(i)

                if grid[j+1][i] == 0: 

                    if digit not in grid[j+1] and digit not in columns[i] and digit not in section: 
                        potential_slots.append(j+1)
                        potential_slots.append(i)

            if len(potential_slots) == 2: 
                grid[potential_slots[0]][potential_slots[1]] = digit 
                columns[potential_slots[1]][potential_slots[0]] = digit 
                


        #Right Sections:
        for j in range(0,6,2): 
            potential_slots = []
            section = grid[j][3:6] + grid[j+1][3:6]
        
            for i in range(3,6):
        
                if grid[j][i] == 0:
        
                    if digit not in grid[j] and digit not in columns[i] and digit not in section:
                        potential_slots.append(j)
                        potential_slots.append(i)
        
                if grid[j+1][i] == 0: 
        
                    if digit not in grid[j+1] and digit not in columns[i] and digit not in section: 
                        potential_slots.append(j+1)
                        potential_slots.append(i)
        
            if len(potential_slots) == 2: 
                grid[potential_slots[0]][potential_slots[1]] = digit 
                columns[potential_slots[1]][potential_slots[0]] = digit 
                

    return grid

def cy_forcer(grid):

    us_record = []
    unfilled_slots = 0


    cycling_mechanism(grid)
    for row in grid:
        for number in row:
            if number == 0:
                unfilled_slots += 1 
    us_record.append(unfilled_slots)
    unfilled_slots = 0 

    cycling_mechanism(grid)
    for row in grid:
        for number in row:
            if number == 0:
                unfilled_slots += 1 
    us_record.append(unfilled_slots)
    unfilled_slots = 0 

   

    if us_record[1] == 0 or us_record[1] == us_record[0]:
        return grid 

    while us_record[-1] != us_record[-2]:

        cycling_mechanism(grid)
        for row in grid:
            for number in row:
                if number == 0:
                    unfilled_slots += 1 
        us_record.append(unfilled_slots)
        unfilled_slots = 0 

    return grid 



def row_fill_mechanism(grid):

    digits = [1,2,3,4,5,6]
    all_numbers = []
    columns = []

    for row in grid:
        for number in row:
            all_numbers.append(number)

    for s in range(0,6):
        column = []
        for d in range(s,36,6):
            column.append(all_numbers[d])
        columns.append(column)

    
    for j in range(0,6):

        for digit in digits:
            row_potential_slots = []

            for i in range(0,6):
                if grid[j][i] == 0 and digit not in grid[j] and digit not in columns[i]:
                    row_potential_slots.append(j)
                    row_potential_slots.append(i)

            if len(row_potential_slots) == 2:
                grid[row_potential_slots[0]][row_potential_slots[1]] = digit 
                columns[row_potential_slots[1]][row_potential_slots[0]] = digit 

    return grid 

def rf_forcer(grid):

    us_record = []
    unfilled_slots = 0

    row_fill_mechanism(grid)
    for row in grid:
        for number in row:
            if number == 0:
                unfilled_slots += 1 
    us_record.append(unfilled_slots)
    unfilled_slots = 0 

    row_fill_mechanism(grid)
    for row in grid:
        for number in row:
            if number == 0:
                unfilled_slots += 1 
    us_record.append(unfilled_slots)
    unfilled_slots = 0 


    if us_record[1] == 0 or us_record[0] == us_record[1]:
        return grid 

    while us_record[-1] != us_record[-2]:

        row_fill_mechanism(grid)
        for row in grid:
            for number in row:
                if number == 0:
                    unfilled_slots += 1 
        us_record.append(unfilled_slots)
        unfilled_slots = 0 

    return grid 



def column_fill_mechanism(grid):

    digits = [1,2,3,4,5,6]
    all_numbers = []
    columns = []
    
    for row in grid:
        for number in row:
            all_numbers.append(number)
    
    for s in range(0,6):
        column = []
        for d in range(s,36,6):
            column.append(all_numbers[d])
        columns.append(column)


    for i in range(0,6):

        for digit in digits: 
            column_potential_slots = []

            for j in range(0,6):
                if columns[i][j] == 0 and digit not in columns[i] and digit not in grid[j]:
                    column_potential_slots.append(i)
                    column_potential_slots.append(j)

            if len(column_potential_slots) == 2:
                grid[column_potential_slots[1]][column_potential_slots[0]] = digit 
                columns[column_potential_slots[0]][column_potential_slots[1]] = digit 

    return grid 

def cf_forcer(grid):

    us_record = []
    unfilled_slots = 0
    
    row_fill_mechanism(grid)
    for row in grid:
         for number in row:
            if number == 0:
                unfilled_slots += 1 
    us_record.append(unfilled_slots)
    unfilled_slots = 0 
    
    row_fill_mechanism(grid)
    for row in grid:
        for number in row:
            if number == 0:
                unfilled_slots += 1 
    us_record.append(unfilled_slots)
    unfilled_slots = 0 
    
    
    if us_record[1] == 0 or us_record[0] == us_record[1]:
        return grid 
    
    while us_record[-1] != us_record[-2]:
    
        row_fill_mechanism(grid)
        for row in grid:
            for number in row:
                if number == 0:
                    unfilled_slots += 1 
        us_record.append(unfilled_slots)
        unfilled_slots = 0 
                    
    return grid 


# DETERMINISTIC GRID GENERATION MECHANICS ^^^ #
###############################################


grid = [[ 1,0,2,0,0,3 ],
        [ 0,0,0,0,0,0 ],
        [ 2,0,4,0,0,0 ],
        [ 0,0,0,4,0,5 ],
        [ 0,0,0,0,0,0 ],
        [ 3,0,0,5,0,1 ]]

grid_preserved = copy.deepcopy(grid)  
  


def grid_solver(grid,grid_preserved):

    us_record = []
    unfilled_slots = 0 

    cy_forcer(grid)
    rf_forcer(grid)
    cf_forcer(grid)
    for row in grid:
        for number in row: 
            if number == 0: 
                unfilled_slots += 1 
    us_record.append(unfilled_slots)
    unfilled_slots = 0 

    cy_forcer(grid)
    rf_forcer(grid)
    cf_forcer(grid)
    for row in grid:
        for number in row: 
            if number == 0: 
                unfilled_slots += 1 
    us_record.append(unfilled_slots)
    unfilled_slots = 0 

    if us_record[1] == 0 or us_record[0] == us_record[1]:

        print("Grid Completely Solved Deterministically:","\n")
        for row in grid:
            print("Row:",row,"\n")
        return "SOLVER TERMINATED"
        
        
    while us_record[-1] != us_record[-2]: 

        cy_forcer(grid)
        rf_forcer(grid)
        cf_forcer(grid)
        for row in grid:
            for number in row: 
                if number == 0:                         
                    unfilled_slots += 1 
        us_record.append(unfilled_slots)
        unfilled_slots = 0 


    grid_populator(grid)

    while column_checker(grid) != 0 or section_checker(grid) != 0: 

        grid = copy.deepcopy(grid_preserved)
        grid_populator(grid)
        column_checker(grid)
        section_checker(grid)

    print("Grid Completely Solved Deterministically And Stochastically;","\n")
    for row in grid:
        print("Row:",row,"\n")
    return "SOLVER TERMINATED"

print(grid_solver(grid,grid_preserved))
```
